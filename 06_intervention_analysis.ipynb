```python

# ================================================================
# PHASE 5: INTERVENTION OPPORTUNITY ANALYSIS
# CORRECTED FINAL VERSION
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display


# ================================================================
# 1. LOAD PHASE 4 AND PHASE 3 OUTPUTS
# ================================================================

project_path = os.path.abspath("..")

processed_path = os.path.join(
    project_path,
    "data",
    "processed"
)

phase4_file = os.path.join(
    processed_path,
    "phase4_analysis_data.csv"
)

phase3_file = os.path.join(
    processed_path,
    "aircraft_connections.csv"
)

if not os.path.exists(phase4_file):
    raise FileNotFoundError(
        "Phase 4 output not found. Run Phase 4 first."
    )

if not os.path.exists(phase3_file):
    raise FileNotFoundError(
        "Phase 3 output not found. Run Phase 3 first."
    )

analysis_data = pd.read_csv(
    phase4_file,
    low_memory=False
)

all_connections = pd.read_csv(
    phase3_file,
    low_memory=False
)

print("=" * 75)
print("PHASE 5: INTERVENTION OPPORTUNITY ANALYSIS")
print("=" * 75)

print("\nPHASE 4 DATA")
print("Rows:", len(analysis_data))
print("Columns:", len(analysis_data.columns))

print("\nPHASE 3 CONNECTION DATA")
print("Rows:", len(all_connections))


# ================================================================
# 2. VALIDATE REQUIRED COLUMNS
# ================================================================

required_phase4_columns = [
    "PREV_FLIGHT_ID",
    "NEXT_FLIGHT_ID",
    "PREV_TAIL_NUM",
    "PREV_CARRIER",
    "PREV_ARR_DELAY",
    "NEXT_DEP_DELAY",
    "ACTUAL_TURNAROUND_MIN",
    "SCHEDULED_TURNAROUND_BUFFER_MIN",
    "BUFFER_CONSUMED_MIN"
]

missing_columns = [
    col for col in required_phase4_columns
    if col not in analysis_data.columns
]

if missing_columns:
    raise ValueError(
        f"Missing Phase 4 columns: {missing_columns}"
    )


required_phase3_columns = [
    "PREV_FLIGHT_ID",
    "NEXT_FLIGHT_ID",
    "PREV_TAIL_NUM",
    "NEXT_DEP_DELAY"
]

missing_phase3_columns = [
    col for col in required_phase3_columns
    if col not in all_connections.columns
]

if missing_phase3_columns:
    raise ValueError(
        f"Missing Phase 3 columns: {missing_phase3_columns}"
    )

print("\nAll required columns are available.")


# ================================================================
# 3. DATA PREPARATION
# ================================================================

numeric_columns = [
    "PREV_ARR_DELAY",
    "NEXT_DEP_DELAY",
    "ACTUAL_TURNAROUND_MIN",
    "SCHEDULED_TURNAROUND_BUFFER_MIN",
    "BUFFER_CONSUMED_MIN"
]

for col in numeric_columns:
    analysis_data[col] = pd.to_numeric(
        analysis_data[col],
        errors="coerce"
    )

phase5_data = analysis_data.dropna(
    subset=[
        "PREV_FLIGHT_ID",
        "NEXT_FLIGHT_ID",
        "PREV_TAIL_NUM",
        "PREV_ARR_DELAY",
        "NEXT_DEP_DELAY",
        "ACTUAL_TURNAROUND_MIN"
    ]
).copy()

phase5_data = phase5_data[
    phase5_data["ACTUAL_TURNAROUND_MIN"] > 0
].copy()

# Normalize identifiers before duplicate checking
for col in [
    "PREV_FLIGHT_ID",
    "NEXT_FLIGHT_ID",
    "PREV_TAIL_NUM"
]:
    phase5_data[col] = phase5_data[col].astype(str).str.strip()

# Remove duplicate connections using aircraft identity too
phase5_data = phase5_data.drop_duplicates(
    subset=[
        "PREV_TAIL_NUM",
        "PREV_FLIGHT_ID",
        "NEXT_FLIGHT_ID"
    ]
).copy()

phase5_data = phase5_data.reset_index(drop=True)

# Estimated indicator, NOT actual remaining turnaround time
phase5_data["ESTIMATED_REMAINING_BUFFER_MIN"] = (
    phase5_data["SCHEDULED_TURNAROUND_BUFFER_MIN"] -
    phase5_data["PREV_ARR_DELAY"]
)

print("\nDATA PREPARATION")
print("-" * 60)

print("Original Phase 4 records:", len(analysis_data))
print("Usable Phase 5 records:", len(phase5_data))

print(
    "Removed records:",
    len(analysis_data) - len(phase5_data)
)

print(
    "Missing scheduled buffer values:",
    phase5_data[
        "SCHEDULED_TURNAROUND_BUFFER_MIN"
    ].isna().sum()
)

print(
    "Missing estimated buffer values:",
    phase5_data[
        "ESTIMATED_REMAINING_BUFFER_MIN"
    ].isna().sum()
)

if phase5_data.empty:
    raise ValueError("No usable connections remain.")


# ================================================================
# 4. IDENTIFY VULNERABLE AIRCRAFT CONNECTIONS
# ================================================================

BASE_DELAY_THRESHOLD = 15
BASE_BUFFER_THRESHOLD = 15


def identify_vulnerable_connections(
    df,
    delay_threshold=15,
    buffer_threshold=15
):

    result = df.copy()

    result["INCOMING_DELAY_MIN"] = (
        result["PREV_ARR_DELAY"].clip(lower=0)
    )

    result["PREVIOUS_DELAYED"] = (
        result["PREV_ARR_DELAY"] >= delay_threshold
    )

    result["LOW_REMAINING_BUFFER"] = (
        result["ESTIMATED_REMAINING_BUFFER_MIN"]
        <= buffer_threshold
    )

    result["DELAYED_WITH_LOW_BUFFER"] = (
        result["PREVIOUS_DELAYED"] &
        result["LOW_REMAINING_BUFFER"]
    )

    result["BUFFER_SHORTFALL"] = (
        result["ESTIMATED_REMAINING_BUFFER_MIN"] < 0
    )

    conditions = [
        (
            result["PREVIOUS_DELAYED"] &
            result["BUFFER_SHORTFALL"]
        ),
        (
            result["PREVIOUS_DELAYED"] &
            result["LOW_REMAINING_BUFFER"]
        ),
        result["PREVIOUS_DELAYED"],
        result["LOW_REMAINING_BUFFER"]
    ]

    labels = [
        "Delayed incoming + buffer shortfall",
        "Delayed incoming + critically low buffer",
        "Delayed incoming + sufficient buffer",
        "On-time incoming + low buffer"
    ]

    result["VULNERABILITY_GROUP"] = np.select(
        conditions,
        labels,
        default="Low immediate vulnerability"
    )

    return result


phase5_data = identify_vulnerable_connections(
    phase5_data,
    delay_threshold=BASE_DELAY_THRESHOLD,
    buffer_threshold=BASE_BUFFER_THRESHOLD
)


print("\nVULNERABILITY GROUP ANALYSIS")
print("-" * 60)

vulnerability_summary = (
    phase5_data
    .groupby("VULNERABILITY_GROUP")
    .agg(
        CONNECTIONS=("NEXT_FLIGHT_ID", "count"),
        AVG_INCOMING_DELAY=("PREV_ARR_DELAY", "mean"),
        AVG_REMAINING_BUFFER=(
            "ESTIMATED_REMAINING_BUFFER_MIN",
            "mean"
        ),
        AVG_NEXT_DEP_DELAY=("NEXT_DEP_DELAY", "mean")
    )
    .reset_index()
    .sort_values("CONNECTIONS", ascending=False)
)

display(vulnerability_summary.round(2))


# ================================================================
# 5. CALCULATE TRANSPARENT INTERVENTION RISK SCORE
# ================================================================

def calculate_intervention_score(
    df,
    delay_threshold=15,
    buffer_threshold=15
):

    result = identify_vulnerable_connections(
        df,
        delay_threshold=delay_threshold,
        buffer_threshold=buffer_threshold
    )

    incoming_delay = (
        result["PREV_ARR_DELAY"].clip(lower=0)
    )

    remaining_buffer = (
        result["ESTIMATED_REMAINING_BUFFER_MIN"]
    )

    scheduled_buffer = (
        result["SCHEDULED_TURNAROUND_BUFFER_MIN"]
    )

    # COMPONENT 1: DELAY SEVERITY (0-30)

    delay_scale = max(
        delay_threshold * 4,
        1
    )

    result["DELAY_SEVERITY_SCORE"] = (
        incoming_delay / delay_scale
    ).clip(upper=1) * 30

    # COMPONENT 2: BUFFER PRESSURE (0-30)

    buffer_unit = max(buffer_threshold, 5)

    buffer_conditions = [
        remaining_buffer <= 0,
        remaining_buffer <= buffer_unit,
        remaining_buffer <= 2 * buffer_unit,
        remaining_buffer <= 4 * buffer_unit
    ]

    buffer_scores = [30, 24, 16, 8]

    result["BUFFER_PRESSURE_SCORE"] = np.select(
        buffer_conditions,
        buffer_scores,
        default=0
    )

    result.loc[
        remaining_buffer.isna(),
        "BUFFER_PRESSURE_SCORE"
    ] = 0

    # COMPONENT 3: SCHEDULED TURNAROUND TIGHTNESS (0-15)

    schedule_conditions = [
        scheduled_buffer <= 30,
        scheduled_buffer <= 60,
        scheduled_buffer <= 90
    ]

    schedule_scores = [15, 10, 5]

    result["SCHEDULE_TIGHTNESS_SCORE"] = np.select(
        schedule_conditions,
        schedule_scores,
        default=0
    )

    result.loc[
        scheduled_buffer.isna(),
        "SCHEDULE_TIGHTNESS_SCORE"
    ] = 0

    # COMPONENT 4: COMBINED DELAY AND BUFFER RISK (0-15)

    result["COMBINED_RISK_SCORE"] = np.where(
        (
            (result["PREV_ARR_DELAY"] >= delay_threshold) &
            (remaining_buffer <= buffer_threshold)
        ),
        15,
        0
    )

    result.loc[
        remaining_buffer.isna(),
        "COMBINED_RISK_SCORE"
    ] = 0

    # FINAL OPERATIONAL RISK SCORE (0-90)

    score_columns = [
        "DELAY_SEVERITY_SCORE",
        "BUFFER_PRESSURE_SCORE",
        "SCHEDULE_TIGHTNESS_SCORE",
        "COMBINED_RISK_SCORE"
    ]

    result["OPERATIONAL_RISK_SCORE"] = (
        result[score_columns].sum(axis=1)
    ).clip(0, 90)

    result["OPERATIONAL_RISK_PERCENT"] = (
        result["OPERATIONAL_RISK_SCORE"] / 90
    ) * 100

    result["INTERVENTION_CATEGORY"] = np.select(
        [
            result["OPERATIONAL_RISK_SCORE"] >= 65,
            result["OPERATIONAL_RISK_SCORE"] >= 35
        ],
        [
            "Priority Review",
            "Review"
        ],
        default="Monitor"
    )

    return result


phase5_data = calculate_intervention_score(
    phase5_data,
    delay_threshold=BASE_DELAY_THRESHOLD,
    buffer_threshold=BASE_BUFFER_THRESHOLD
)


# ================================================================
# 6. TRACE MULTI-FLIGHT CHAINS AND DOWNSTREAM EXPOSURE
# CORRECTED: SAME-AIRCRAFT FLIGHT LOOKUP
# ================================================================

graph_data = all_connections.copy()

# Drop missing identifiers BEFORE converting them to strings
graph_data = graph_data.dropna(
    subset=[
        "PREV_TAIL_NUM",
        "PREV_FLIGHT_ID",
        "NEXT_FLIGHT_ID"
    ]
).copy()

for col in [
    "PREV_TAIL_NUM",
    "PREV_FLIGHT_ID",
    "NEXT_FLIGHT_ID"
]:
    graph_data[col] = (
        graph_data[col].astype(str).str.strip()
    )

graph_data["NEXT_DEP_DELAY"] = pd.to_numeric(
    graph_data["NEXT_DEP_DELAY"],
    errors="coerce"
)

# Each aircraft and previous flight should have one outgoing connection
duplicate_starts = graph_data.duplicated(
    subset=["PREV_TAIL_NUM", "PREV_FLIGHT_ID"]
).sum()

print("\nDuplicate aircraft-flight starting points:", duplicate_starts)

graph_data = graph_data.drop_duplicates(
    subset=["PREV_TAIL_NUM", "PREV_FLIGHT_ID"],
    keep="first"
).copy()

# Create lookup using aircraft registration + flight ID
connection_lookup = {
    (row.PREV_TAIL_NUM, row.PREV_FLIGHT_ID): {
        "NEXT_FLIGHT_ID": row.NEXT_FLIGHT_ID,
        "NEXT_DEP_DELAY": row.NEXT_DEP_DELAY
    }
    for row in graph_data.itertuples(index=False)
}


def calculate_downstream_exposure(
    row,
    connection_lookup,
    delay_threshold=15,
    max_chain_length=10
):

    aircraft = row["PREV_TAIL_NUM"]

    current_flight = row["NEXT_FLIGHT_ID"]

    visited_flights = {
        row["PREV_FLIGHT_ID"],
        row["NEXT_FLIGHT_ID"]
    }

    downstream_count = 0
    downstream_delayed_count = 0
    downstream_delay_total = 0
    maximum_downstream_delay = 0

    for step in range(max_chain_length):

        lookup_key = (
            aircraft,
            current_flight
        )

        next_connection = connection_lookup.get(
            lookup_key
        )

        if next_connection is None:
            break

        downstream_flight = str(
            next_connection["NEXT_FLIGHT_ID"]
        )

        if downstream_flight in visited_flights:
            break

        visited_flights.add(downstream_flight)

        downstream_delay = next_connection[
            "NEXT_DEP_DELAY"
        ]

        downstream_count += 1

        if pd.notna(downstream_delay):

            if downstream_delay >= delay_threshold:
                downstream_delayed_count += 1

            positive_delay = max(
                float(downstream_delay),
                0
            )

            downstream_delay_total += positive_delay

            maximum_downstream_delay = max(
                maximum_downstream_delay,
                positive_delay
            )

        current_flight = downstream_flight

    if downstream_count > 0:

        downstream_delay_rate = (
            downstream_delayed_count /
            downstream_count
        ) * 100

    else:

        downstream_delay_rate = np.nan

    return pd.Series({

        "DOWNSTREAM_CHAIN_LENGTH": downstream_count,

        "DOWNSTREAM_DELAYED_FLIGHTS": downstream_delayed_count,

        "DOWNSTREAM_DELAY_RATE": downstream_delay_rate,

        "DOWNSTREAM_DELAY_MINUTES": downstream_delay_total,

        "MAX_DOWNSTREAM_DELAY_MIN": maximum_downstream_delay,

        "DOWNSTREAM_DATA_AVAILABLE": downstream_count > 0

    })


print("\nCalculating corrected downstream exposure...")

downstream_results = phase5_data.apply(
    calculate_downstream_exposure,
    axis=1,
    connection_lookup=connection_lookup,
    delay_threshold=BASE_DELAY_THRESHOLD,
    max_chain_length=10
)

phase5_data = phase5_data.drop(
    columns=[
        col for col in [
            "DOWNSTREAM_CHAIN_LENGTH",
            "DOWNSTREAM_DELAYED_FLIGHTS",
            "DOWNSTREAM_DELAY_RATE",
            "DOWNSTREAM_DELAY_MINUTES",
            "MAX_DOWNSTREAM_DELAY_MIN",
            "DOWNSTREAM_DATA_AVAILABLE"
        ]
        if col in phase5_data.columns
    ]
)

phase5_data = pd.concat(
    [
        phase5_data.reset_index(drop=True),
        downstream_results.reset_index(drop=True)
    ],
    axis=1
)

print("\nCorrected downstream tracing completed.")

print("\nDOWNSTREAM EXPOSURE SUMMARY")

display(
    phase5_data[
        [
            "PREV_FLIGHT_ID",
            "NEXT_FLIGHT_ID",
            "PREV_TAIL_NUM",
            "DOWNSTREAM_CHAIN_LENGTH",
            "DOWNSTREAM_DELAYED_FLIGHTS",
            "DOWNSTREAM_DELAY_RATE",
            "DOWNSTREAM_DELAY_MINUTES"
        ]
    ].head(10).round(2)
)

print("\nVALIDATION")

print("Connections analyzed:", len(phase5_data))

print(
    "Connections with downstream flights:",
    phase5_data["DOWNSTREAM_DATA_AVAILABLE"].sum()
)

print(
    "Maximum traced chain length:",
    phase5_data["DOWNSTREAM_CHAIN_LENGTH"].max()
)


# ================================================================
# 7. CALCULATE COMBINED INTERVENTION PRIORITY
# ================================================================

if phase5_data["DOWNSTREAM_DELAY_MINUTES"].nunique() > 1:

    phase5_data["DOWNSTREAM_EXPOSURE_PERCENTILE"] = (
        phase5_data["DOWNSTREAM_DELAY_MINUTES"]
        .rank(pct=True, method="average") * 100
    )

else:

    phase5_data["DOWNSTREAM_EXPOSURE_PERCENTILE"] = 0


phase5_data["INTERVENTION_PRIORITY_SCORE"] = (
    0.80 * phase5_data["OPERATIONAL_RISK_PERCENT"] +
    0.20 * phase5_data["DOWNSTREAM_EXPOSURE_PERCENTILE"]
)

phase5_data = phase5_data.sort_values(
    "INTERVENTION_PRIORITY_SCORE",
    ascending=False
).reset_index(drop=True)

phase5_data["INTERVENTION_PRIORITY_RANK"] = (
    np.arange(1, len(phase5_data) + 1)
)


# ================================================================
# 8. CREATE FINAL INTERVENTION-PRIORITY FRAMEWORK
# ================================================================

tier_order = {
    "Priority Review": 1,
    "Review": 2,
    "Monitor": 3
}

phase5_data["TIER_ORDER"] = (
    phase5_data["INTERVENTION_CATEGORY"].map(tier_order)
)

phase5_data = phase5_data.sort_values(
    [
        "TIER_ORDER",
        "INTERVENTION_PRIORITY_SCORE"
    ],
    ascending=[True, False]
).reset_index(drop=True)

phase5_data["INTERVENTION_PRIORITY_RANK"] = (
    np.arange(1, len(phase5_data) + 1)
)


def generate_intervention_recommendation(row):

    recommendations = []

    if row["PREV_ARR_DELAY"] >= 30:
        recommendations.append(
            "Review incoming flight delay"
        )

    if row["ESTIMATED_REMAINING_BUFFER_MIN"] <= 0:
        recommendations.append(
            "Urgent turnaround feasibility review"
        )

    elif row["ESTIMATED_REMAINING_BUFFER_MIN"] <= 15:
        recommendations.append(
            "Check ground turnaround readiness"
        )

    if row["SCHEDULED_TURNAROUND_BUFFER_MIN"] <= 30:
        recommendations.append(
            "Review tight scheduled connection"
        )

    if row["DOWNSTREAM_DELAYED_FLIGHTS"] >= 2:
        recommendations.append(
            "Monitor downstream aircraft rotation"
        )

    if not recommendations:
        recommendations.append(
            "Continue routine monitoring"
        )

    return "; ".join(recommendations)


phase5_data["SUGGESTED_INTERVENTION"] = (
    phase5_data.apply(
        generate_intervention_recommendation,
        axis=1
    )
)


# ================================================================
# 9. INTERVENTION TIER SUMMARY
# ================================================================

tier_summary = (
    phase5_data
    .groupby("INTERVENTION_CATEGORY")
    .agg(
        CONNECTIONS=("NEXT_FLIGHT_ID", "count"),
        AVG_OPERATIONAL_RISK=(
            "OPERATIONAL_RISK_SCORE",
            "mean"
        ),
        AVG_PRIORITY_SCORE=(
            "INTERVENTION_PRIORITY_SCORE",
            "mean"
        ),
        AVG_INCOMING_DELAY=(
            "PREV_ARR_DELAY",
            "mean"
        ),
        AVG_REMAINING_BUFFER=(
            "ESTIMATED_REMAINING_BUFFER_MIN",
            "mean"
        ),
        AVG_DOWNSTREAM_DELAY_MINUTES=(
            "DOWNSTREAM_DELAY_MINUTES",
            "mean"
        ),
        TOTAL_DOWNSTREAM_DELAYED_FLIGHTS=(
            "DOWNSTREAM_DELAYED_FLIGHTS",
            "sum"
        )
    )
    .reset_index()
)

tier_summary["PERCENT_OF_CONNECTIONS"] = (
    tier_summary["CONNECTIONS"] /
    len(phase5_data) * 100
)

print("\nINTERVENTION TIER SUMMARY")

display(tier_summary.round(2))


# ================================================================
# 10. RANK POTENTIAL INTERVENTION POINTS
# ================================================================

ranking_columns = [
    "INTERVENTION_PRIORITY_RANK",
    "PREV_FLIGHT_ID",
    "NEXT_FLIGHT_ID",
    "PREV_TAIL_NUM",
    "PREV_CARRIER",
    "PREV_ARR_DELAY",
    "SCHEDULED_TURNAROUND_BUFFER_MIN",
    "ESTIMATED_REMAINING_BUFFER_MIN",
    "OPERATIONAL_RISK_SCORE",
    "DOWNSTREAM_CHAIN_LENGTH",
    "DOWNSTREAM_DELAYED_FLIGHTS",
    "DOWNSTREAM_DELAY_MINUTES",
    "INTERVENTION_PRIORITY_SCORE",
    "INTERVENTION_CATEGORY",
    "SUGGESTED_INTERVENTION"
]

for col in [
    "PREV_ORIGIN",
    "PREV_DEST",
    "NEXT_ORIGIN",
    "NEXT_DEST"
]:
    if col in phase5_data.columns:
        ranking_columns.insert(4, col)

ranking_columns = [
    col for col in ranking_columns
    if col in phase5_data.columns
]

ranked_intervention_points = (
    phase5_data[ranking_columns]
    .sort_values("INTERVENTION_PRIORITY_RANK")
    .copy()
)

print("\nTOP 20 INTERVENTION POINTS")

display(
    ranked_intervention_points.head(20).round(2)
)


# ================================================================
# 11. THRESHOLD SENSITIVITY ANALYSIS
# CORRECTED: STABLE CONNECTION IDENTIFIERS
# ================================================================

delay_thresholds = [10, 15, 30]
buffer_thresholds = [0, 15, 30]

# Create a stable identifier for each aircraft connection
phase5_data["CONNECTION_KEY"] = (
    phase5_data["PREV_TAIL_NUM"].astype(str) + "_" +
    phase5_data["PREV_FLIGHT_ID"].astype(str) + "_" +
    phase5_data["NEXT_FLIGHT_ID"].astype(str)
)

if phase5_data["CONNECTION_KEY"].duplicated().any():
    raise ValueError("Duplicate connection keys found.")

top_k = min(10, len(phase5_data))

baseline_top_keys = set(
    phase5_data.nlargest(
        top_k,
        "INTERVENTION_PRIORITY_SCORE"
    )["CONNECTION_KEY"]
)

baseline_ranks = (
    phase5_data
    .set_index("CONNECTION_KEY")["INTERVENTION_PRIORITY_SCORE"]
    .rank(ascending=False, method="average")
)

sensitivity_results = []


for delay_threshold in delay_thresholds:

    for buffer_threshold in buffer_thresholds:

        scenario = calculate_intervention_score(
            phase5_data,
            delay_threshold=delay_threshold,
            buffer_threshold=buffer_threshold
        )

        scenario["SCENARIO_PRIORITY_SCORE"] = (
            0.80 * scenario["OPERATIONAL_RISK_PERCENT"] +
            0.20 * scenario["DOWNSTREAM_EXPOSURE_PERCENTILE"]
        )

        scenario_top_keys = set(
            scenario.nlargest(
                top_k,
                "SCENARIO_PRIORITY_SCORE"
            )["CONNECTION_KEY"]
        )

        overlap = len(
            baseline_top_keys.intersection(scenario_top_keys)
        )

        scenario_ranks = (
            scenario
            .set_index("CONNECTION_KEY")["SCENARIO_PRIORITY_SCORE"]
            .rank(ascending=False, method="average")
        )

        rank_correlation = scenario_ranks.corr(
            baseline_ranks,
            method="spearman"
        )

        tier_counts = (
            scenario["INTERVENTION_CATEGORY"].value_counts()
        )

        sensitivity_results.append({

            "DELAY_THRESHOLD_MIN": delay_threshold,

            "BUFFER_THRESHOLD_MIN": buffer_threshold,

            "PRIORITY_REVIEW_COUNT": tier_counts.get(
                "Priority Review", 0
            ),

            "REVIEW_COUNT": tier_counts.get(
                "Review", 0
            ),

            "MONITOR_COUNT": tier_counts.get(
                "Monitor", 0
            ),

            "TOP_10_OVERLAP_WITH_BASELINE": overlap,

            "TOP_10_OVERLAP_PERCENT": (
                overlap / max(top_k, 1)
            ) * 100,

            "SPEARMAN_RANK_CORRELATION": rank_correlation

        })


threshold_sensitivity = pd.DataFrame(
    sensitivity_results
)

print("\nTHRESHOLD SENSITIVITY ANALYSIS")

display(
    threshold_sensitivity.round(3)
)


# ================================================================
# 12. IDENTIFY MOST STABLE INTERVENTION POINTS
# CORRECTED: TRACK USING CONNECTION KEY
# ================================================================

top_connection_frequency = {}

for delay_threshold in delay_thresholds:

    for buffer_threshold in buffer_thresholds:

        scenario = calculate_intervention_score(
            phase5_data,
            delay_threshold=delay_threshold,
            buffer_threshold=buffer_threshold
        )

        scenario["SCENARIO_PRIORITY_SCORE"] = (
            0.80 * scenario["OPERATIONAL_RISK_PERCENT"] +
            0.20 * scenario["DOWNSTREAM_EXPOSURE_PERCENTILE"]
        )

        top_keys = scenario.nlargest(
            top_k,
            "SCENARIO_PRIORITY_SCORE"
        )["CONNECTION_KEY"]

        for key in top_keys:

            top_connection_frequency[key] = (
                top_connection_frequency.get(key, 0) + 1
            )


phase5_data["TOP_10_SCENARIO_FREQUENCY"] = (
    phase5_data["CONNECTION_KEY"]
    .map(top_connection_frequency)
    .fillna(0)
    .astype(int)
)

phase5_data["TOP_10_SCENARIO_STABILITY_PERCENT"] = (
    phase5_data["TOP_10_SCENARIO_FREQUENCY"] /
    (len(delay_thresholds) * len(buffer_thresholds))
) * 100


stable_intervention_points = (
    phase5_data
    .sort_values(
        [
            "TOP_10_SCENARIO_STABILITY_PERCENT",
            "INTERVENTION_PRIORITY_SCORE"
        ],
        ascending=False
    )
    .copy()
)

print("\nMOST STABLE INTERVENTION POINTS")

display(
    stable_intervention_points[
        [
            "PREV_FLIGHT_ID",
            "NEXT_FLIGHT_ID",
            "PREV_TAIL_NUM",
            "OPERATIONAL_RISK_SCORE",
            "DOWNSTREAM_DELAY_MINUTES",
            "INTERVENTION_PRIORITY_SCORE",
            "TOP_10_SCENARIO_FREQUENCY",
            "TOP_10_SCENARIO_STABILITY_PERCENT",
            "INTERVENTION_CATEGORY"
        ]
    ].head(20).round(2)
)


# ================================================================
# 13. VISUALIZE INTERVENTION CATEGORIES
# ================================================================

tier_plot = (
    phase5_data["INTERVENTION_CATEGORY"]
    .value_counts()
    .reindex(
        [
            "Priority Review",
            "Review",
            "Monitor"
        ],
        fill_value=0
    )
)

plt.figure(figsize=(8, 5))

plt.bar(
    tier_plot.index,
    tier_plot.values
)

plt.title(
    "Aircraft Connections by Intervention Category"
)

plt.xlabel("Intervention Category")
plt.ylabel("Number of Connections")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()


# ================================================================
# 14. VISUALIZE TOP INTERVENTION POINTS
# ================================================================

top_points = (
    ranked_intervention_points
    .head(15)
    .copy()
)

plt.figure(figsize=(11, 6))

plt.barh(
    top_points["NEXT_FLIGHT_ID"].astype(str),
    top_points["INTERVENTION_PRIORITY_SCORE"]
)

plt.gca().invert_yaxis()

plt.title(
    "Top 15 Aircraft Connection Intervention Priorities"
)

plt.xlabel("Intervention Priority Score")
plt.ylabel("Next Flight ID")

plt.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()


# ================================================================
# 15. VISUALIZE OPERATIONAL RISK VS DOWNSTREAM EXPOSURE
# ================================================================

plt.figure(figsize=(9, 6))

plt.scatter(
    phase5_data["OPERATIONAL_RISK_SCORE"],
    phase5_data["DOWNSTREAM_DELAY_MINUTES"],
    alpha=0.35,
    s=15
)

plt.title(
    "Operational Risk vs Historical Downstream Delay Exposure"
)

plt.xlabel("Operational Risk Score")
plt.ylabel("Downstream Delay Exposure (Minutes)")

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ================================================================
# 16. FINAL PHASE 5 SUMMARY
# ================================================================

total_connections = len(phase5_data)

priority_count = (
    phase5_data["INTERVENTION_CATEGORY"] ==
    "Priority Review"
).sum()

review_count = (
    phase5_data["INTERVENTION_CATEGORY"] ==
    "Review"
).sum()

monitor_count = (
    phase5_data["INTERVENTION_CATEGORY"] ==
    "Monitor"
).sum()

print("\n" + "=" * 75)
print("PHASE 5: FINAL INTERVENTION OPPORTUNITY SUMMARY")
print("=" * 75)

print("\nDATASET")
print("Total analyzed connections:", total_connections)

print("\nINTERVENTION CATEGORIES")

print(
    "Priority Review:",
    priority_count,
    f"({priority_count / total_connections * 100:.2f}%)"
)

print(
    "Review:",
    review_count,
    f"({review_count / total_connections * 100:.2f}%)"
)

print(
    "Monitor:",
    monitor_count,
    f"({monitor_count / total_connections * 100:.2f}%)"
)

print("\nOPERATIONAL RISK")

print(
    "Average operational risk score:",
    round(phase5_data["OPERATIONAL_RISK_SCORE"].mean(), 2)
)

print(
    "Maximum operational risk score:",
    round(phase5_data["OPERATIONAL_RISK_SCORE"].max(), 2)
)

print("\nDOWNSTREAM EXPOSURE")

print(
    "Connections with observed downstream flights:",
    phase5_data["DOWNSTREAM_DATA_AVAILABLE"].sum()
)

print(
    "Total downstream delayed flights:",
    phase5_data["DOWNSTREAM_DELAYED_FLIGHTS"].sum()
)

print(
    "Total downstream positive delay minutes:",
    round(
        phase5_data["DOWNSTREAM_DELAY_MINUTES"].sum(),
        2
    )
)

print("\nTHRESHOLD SENSITIVITY")

print(
    "Number of tested threshold combinations:",
    len(threshold_sensitivity)
)

print(
    "Average top-10 overlap with baseline:",
    round(
        threshold_sensitivity[
            "TOP_10_OVERLAP_PERCENT"
        ].mean(),
        2
    ),
    "%"
)


# ================================================================
# 17. SAVE ALL PHASE 5 OUTPUTS
# ================================================================

os.makedirs(
    processed_path,
    exist_ok=True
)

phase5_data.to_csv(
    os.path.join(
        processed_path,
        "phase5_intervention_analysis.csv"
    ),
    index=False
)

ranked_intervention_points.to_csv(
    os.path.join(
        processed_path,
        "phase5_ranked_intervention_points.csv"
    ),
    index=False
)

vulnerability_summary.to_csv(
    os.path.join(
        processed_path,
        "phase5_vulnerability_summary.csv"
    ),
    index=False
)

tier_summary.to_csv(
    os.path.join(
        processed_path,
        "phase5_intervention_tier_summary.csv"
    ),
    index=False
)

threshold_sensitivity.to_csv(
    os.path.join(
        processed_path,
        "phase5_threshold_sensitivity.csv"
    ),
    index=False
)

stable_intervention_points.to_csv(
    os.path.join(
        processed_path,
        "phase5_stable_intervention_points.csv"
    ),
    index=False
)

print("\n" + "=" * 75)
print("PHASE 5 OUTPUTS SAVED SUCCESSFULLY")
print("=" * 75)

output_files = [
    "phase5_intervention_analysis.csv",
    "phase5_ranked_intervention_points.csv",
    "phase5_vulnerability_summary.csv",
    "phase5_intervention_tier_summary.csv",
    "phase5_threshold_sensitivity.csv",
    "phase5_stable_intervention_points.csv"
]

for filename in output_files:
    print(filename)

print("\nFinal Phase 5 dataset shape:", phase5_data.shape)

display(
    ranked_intervention_points.head(10).round(2)
)
```
