```python
def convert_hhmm(value):
    if pd.isna(value):
        return pd.NaT

    value = int(value)

    hours = value // 100
    minutes = value % 100

    return pd.Timedelta(hours=hours, minutes=minutes)
```

```python
for col in ["DEP_TIME", "ARR_TIME", "CRS_DEP_TIME", "CRS_ARR_TIME"]:
    rotation_flights[col + "_TD"] = rotation_flights[col].apply(convert_hhmm)
```

```python
import airportsdata

airports = airportsdata.load("IATA")

print(airports["BOS"])
```

```python
airport_codes = set(rotation_flights["ORIGIN"].dropna()) | set(rotation_flights["DEST"].dropna())

airport_tz = {}

for code in airport_codes:
    if code in airports:
        airport_tz[code] = airports[code]["tz"]

print("Total airports:", len(airport_tz))
print(airport_tz)
```

```python


airport_tz["PBI"] = "America/New_York"
missing_airports = airport_codes - set(airport_tz.keys())

print("Missing airports:", missing_airports)
print("Missing count:", len(missing_airports))
```

```python
from zoneinfo import ZoneInfo

def convert_to_utc(row, time_col, airport_col):
    local_time = row[time_col]
    airport = row[airport_col]

    if pd.isna(local_time) or pd.isna(airport):
        return pd.NaT

    timezone = ZoneInfo(airport_tz[airport])

    return (
        local_time
        .to_pydatetime()
        .replace(tzinfo=timezone)
        .astimezone(ZoneInfo("UTC"))
    )

```

```python
rotation_flights["DEP_UTC"] = rotation_flights.apply(
    lambda row: convert_to_utc(row, "DEP_TIMESTAMP", "ORIGIN"),
    axis=1
)

rotation_flights["ARR_UTC"] = rotation_flights.apply(
    lambda row: convert_to_utc(row, "ARR_TIMESTAMP", "DEST"),
    axis=1
)
```

```python
rotation_flights["DEP_LOCAL"] = (
    rotation_flights["FL_DATE"] + rotation_flights["DEP_TIME_TD"]
)

rotation_flights["ARR_LOCAL"] = (
    rotation_flights["FL_DATE"] + rotation_flights["ARR_TIME_TD"]
)
```

```python
rotation_flights["DEP_UTC"] = pd.Series(
    pd.NaT, index=rotation_flights.index, dtype="datetime64[ns, UTC]"
)

rotation_flights["ARR_UTC"] = pd.Series(
    pd.NaT, index=rotation_flights.index, dtype="datetime64[ns, UTC]"
)

for airport, tz in airport_tz.items():

    dep_mask = rotation_flights["ORIGIN"] == airport
    arr_mask = rotation_flights["DEST"] == airport

    rotation_flights.loc[dep_mask, "DEP_UTC"] = (
        rotation_flights.loc[dep_mask, "DEP_LOCAL"]
        .dt.tz_localize(
            tz, ambiguous="NaT", nonexistent="shift_forward"
        )
        .dt.tz_convert("UTC")
    )

    rotation_flights.loc[arr_mask, "ARR_UTC"] = (
        rotation_flights.loc[arr_mask, "ARR_LOCAL"]
        .dt.tz_localize(
            tz, ambiguous="NaT", nonexistent="shift_forward"
        )
        .dt.tz_convert("UTC")
    )
```

```python
overnight_mask = (
    rotation_flights["ARR_UTC"] <= rotation_flights["DEP_UTC"]
)

rotation_flights.loc[overnight_mask, "ARR_UTC"] += pd.Timedelta(days=1)
```

```python
rotation_flights["FLIGHT_DURATION"] = (
    rotation_flights["ARR_UTC"] - rotation_flights["DEP_UTC"]
)
```

```python
print("Flight duration statistics:")
print(rotation_flights["FLIGHT_DURATION"].describe())

print("\nNegative durations:",
      (rotation_flights["FLIGHT_DURATION"] < pd.Timedelta(0)).sum())

print("Durations above 24 hours:",
      (rotation_flights["FLIGHT_DURATION"] > pd.Timedelta(hours=24)).sum())

print("Missing departure UTC:", rotation_flights["DEP_UTC"].isna().sum())

print("Missing arrival UTC:", rotation_flights["ARR_UTC"].isna().sum())
```

```python
# Identify flights travelling from Guam to Honolulu
dateline_mask = (
    (rotation_flights["ORIGIN"] == "GUM") &
    (rotation_flights["DEST"] == "HNL")
)

# Arrival in Honolulu is on the previous local calendar day
rotation_flights.loc[dateline_mask, "ARR_LOCAL"] -= pd.Timedelta(days=1)

# Recalculate their arrival timestamps in UTC
rotation_flights.loc[dateline_mask, "ARR_UTC"] = (
    rotation_flights.loc[dateline_mask, "ARR_LOCAL"]
    .dt.tz_localize("Pacific/Honolulu")
    .dt.tz_convert("UTC")
)

# Recalculate flight duration
rotation_flights["FLIGHT_DURATION"] = (
    rotation_flights["ARR_UTC"] - rotation_flights["DEP_UTC"]
)
```
